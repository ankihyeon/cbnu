## ex2. OpenCV를 활용하여 동영상 입출력 - 동영상 파일 (day3/ex2_opencv_readVideo.py)
test_video.mp4 동영상을 읽고 출력

```python
import cv2

# 동영상 파일 열기
video_path = 'test_video.mp4'
cap = cv2.VideoCapture(video_path)

# 동영상 파일 열기 확인
if not cap.isOpened():
    print(f'동영상 파일을 열 수 없습니다: {video_path}')
else:
    # 동영상 프레임 읽고 출력
    while True:
        ret, frame = cap.read()  # 동영상 프레임 읽기
        if not ret:
            print('동영상 재생이 끝났습니다.')
            break

        cv2.imshow('Video Playback', frame)  # 프레임 출력

        # 'q' 키를 누르면 종료
        if cv2.waitKey(25) & 0xFF == ord('q'):
            break

    # 자원 해제
    cap.release()  # 비디오 캡처 객체 닫기
    cv2.destroyAllWindows()  # 모든 창 닫기
```

## ex6. 색상 영역 검출 - RGB 색 공간 활용 (day3/ex6_opencv_extractColor(RGB).py)
candies.png 이미지에서 Red 성분이 50 이상인 색만 추출

```python
import cv2
import numpy as np

# 이미지 읽기
image = cv2.imread('candies.png')

# 이미지 읽기가 성공적이지 못한 경우 처리
if image is None:
    print('이미지를 열 수 없습니다. 파일 경로를 확인하세요.')
else:
    # BGR 채널 분리
    blue, green, red = cv2.split(image)

    # Red 성분이 50 이상인 영역(마스크) 생성
    red_mask = cv2.inRange(red, 50, 255)

    # 마스크를 적용하여 원래 이미지의 해당 영역 추출
    filtered_image = cv2.bitwise_and(image, image, mask=red_mask)

    # 결과 이미지와 마스크 출력
    cv2.imshow('Original Image', image)
    cv2.imshow('Red >= 50 Mask', red_mask)
    cv2.imshow('Filtered Image', filtered_image)

    # 키 입력 대기 후 창 닫기
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

## ex6 (HSV). 색상 영역 검출 - HSV 색 공간 활용 (빨간색 캔디 추출)
candies.png 이미지에서 붉은색 캔디만 추출 (HSV 색 공간 활용)

```python
import cv2
import numpy as np

# 이미지 읽기
image = cv2.imread('candies.png')

if image is None:
    print('이미지를 열 수 없습니다. 파일 경로를 확인하세요.')
else:
    # BGR 이미지를 HSV 색 공간으로 변환
    hsv_image = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)

    # 붉은색 범위 지정 (HSV 범위)
    # OpenCV에서 빨간색은 두 영역으로 나뉘어 있음 (0-10과 170-180)
    lower_red1 = np.array([0, 120, 70])    # 첫 번째 빨간색 범위 (Hue: 0-10)
    upper_red1 = np.array([10, 255, 255])
    lower_red2 = np.array([170, 120, 70])  # 두 번째 빨간색 범위 (Hue: 170-180)
    upper_red2 = np.array([180, 255, 255])

    # 각각의 범위에 대해 마스크 생성
    mask1 = cv2.inRange(hsv_image, lower_red1, upper_red1)
    mask2 = cv2.inRange(hsv_image, lower_red2, upper_red2)

    # 두 마스크를 합침
    red_mask = cv2.bitwise_or(mask1, mask2)

    # 마스크를 원본 이미지에 적용하여 빨간색만 필터링
    red_candies = cv2.bitwise_and(image, image, mask=red_mask)

    cv2.imshow('Red Mask', red_mask)
    cv2.imshow('Red Candies', red_candies)

    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

## [실습] candies.png 이미지에서 파란색 캔디만 추출하기 (HSV 색 공간 활용)
[힌트] 파란색 영역은 (100-140) 임

```python
import cv2
import numpy as np

# 이미지 읽기
image = cv2.imread('candies.png')

if image is None:
    print('이미지를 열 수 없습니다. 파일 경로를 확인하세요.')
else:
    # BGR 이미지를 HSV 색 공간으로 변환
    hsv_image = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)

    # 파란색 범위 지정 (Hue: 100-140)
    lower_blue = np.array([100, 50, 50])
    upper_blue = np.array([140, 255, 255])

    # 마스크 생성
    blue_mask = cv2.inRange(hsv_image, lower_blue, upper_blue)

    # 마스크를 원본 이미지에 적용
    blue_candies = cv2.bitwise_and(image, image, mask=blue_mask)

    cv2.imshow('Blue Mask', blue_mask)
    cv2.imshow('Blue Candies', blue_candies)

    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

## ex7. 히스토그램 - Grayscale (day3/ex7_opencv_histogram(grayscale).py)
Lenna.png 파일을 Grayscale로 읽고, 히스토그램 출력

```python
import cv2
import matplotlib.pyplot as plt

# Lenna 이미지를 그레이스케일로 읽기
image = cv2.imread('Lenna.png', cv2.IMREAD_GRAYSCALE)

# 이미지 읽기가 성공적이지 못한 경우 처리
if image is None:
    print('이미지를 열 수 없습니다. 파일 경로를 확인하세요.')
else:
    # 히스토그램 계산
    hist = cv2.calcHist([image], [0], None, [256], [0, 256])

    # 원본 이미지 출력
    cv2.imshow('Grayscale Image', image)

    # 히스토그램 출력
    plt.figure()
    plt.title('Grayscale Histogram')
    plt.xlabel('Pixel Value')
    plt.ylabel('Frequency')
    plt.plot(hist)
    plt.xlim([0, 256])
    plt.show()

    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

## ex7 (BGR). 히스토그램 - BGR
Lenna.png 파일을 BGR로 읽고, 히스토그램 출력

```python
import cv2
import matplotlib.pyplot as plt

# Lenna 이미지를 BGR로 읽기
image = cv2.imread('Lenna.png')

if image is None:
    print('이미지를 열 수 없습니다. 파일 경로를 확인하세요.')
else:
    colors = ('b', 'g', 'r')
    labels = ('Blue', 'Green', 'Red')

    # 원본 이미지 출력
    cv2.imshow('BGR Image', image)

    # BGR 히스토그램 출력
    plt.figure()
    plt.title('BGR Histogram')
    plt.xlabel('Pixel Value')
    plt.ylabel('Frequency')

    for i, (color, label) in enumerate(zip(colors, labels)):
        hist = cv2.calcHist([image], [i], None, [256], [0, 256])
        plt.plot(hist, color=color, label=label)

    plt.xlim([0, 256])
    plt.legend()
    plt.show()

    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

## ex8. 명암비 조작 (day3/ex8_opencv_adjustWBR.py)
Lenna.png 흑백 이미지의 명암비 조절

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread('Lenna.png', cv2.IMREAD_GRAYSCALE)

if image is None:
    print('이미지를 열 수 없습니다. 파일 경로를 확인하세요.')
else:
    alpha = 1.0
    func = (1 + alpha) * image - (alpha * 128)
    dst = np.clip(func, 0, 255).astype(np.uint8)

    cv2.imshow('Original Grayscale Image', image)
    cv2.imshow('Contrast Adjusted Image', dst)

    # 히스토그램 비교
    plt.figure(figsize=(12, 4))

    plt.subplot(1, 2, 1)
    plt.title('Histogram of Original Image')
    plt.xlabel('Pixel Intensity')
    plt.ylabel('Frequency')
    hist_orig = cv2.calcHist([image], [0], None, [256], [0, 256])
    plt.plot(hist_orig, color='blue')

    plt.subplot(1, 2, 2)
    plt.title('Histogram of Contrast Adjusted Image')
    plt.xlabel('Pixel Intensity')
    plt.ylabel('Frequency')
    hist_dst = cv2.calcHist([dst], [0], None, [256], [0, 256])
    plt.plot(hist_dst, color='red')

    plt.tight_layout()
    plt.show()

    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

## ex8 (equalizeHist). 히스토그램 평활화 (균등화) (day3/ex8_equalizeHist.py)
Hawkes.jpg 이미지에 히스토그램 평활화 적용

## ex8 (backproj). 히스토그램 역투영 (day3/ex8_opencv_backproj.py)
desert.JPG 이미지에 히스토그램 역투영 적용

```python
import cv2
import numpy as np

src = cv2.imread('desert.JPG', cv2.IMREAD_COLOR)

if src is None:
    print('이미지를 열 수 없습니다. 파일 경로를 확인하세요.')
else:
    x, y, w, h = cv2.selectROI(src)
    src_ycrcb = cv2.cvtColor(src, cv2.COLOR_BGR2YCrCb)
    roi = src_ycrcb[y:y+h, x:x+w]
    hist = cv2.calcHist([roi], [1, 2], None, [128, 128], [0, 256, 0, 256])
    backproj = cv2.calcBackProject([src_ycrcb], [1, 2], hist, [0, 256, 0, 256], 1)
    dst = cv2.copyTo(src, backproj)

    cv2.imshow('BackProject', dst)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

## ex9. 이미지 필터링 (day3/ex9_opencv_filtering.py)
Lenna.png 이미지에 평균값 필터, 샤프닝 필터, 라플라시안 필터 적용

```python
import cv2
import numpy as np

image = cv2.imread('Lenna.png')

if image is None:
    print('이미지를 열 수 없습니다. 파일 경로를 확인하세요.')
else:
    # 2D 필터 적용 함수
    def apply_filter(image, kernel, filter_name):
        filtered_image = cv2.filter2D(image, -1, kernel)
        cv2.imshow(f'{filter_name}', filtered_image)
        cv2.imwrite(f'{filter_name}.jpg', filtered_image)  # 필터 적용된 결과 저장
        return filtered_image

    # 1. 평균 값 필터(3x3)
    average_filter = np.ones((3, 3), np.float32) / 9.0

    # 2. 샤프닝 필터(3x3)
    sharpening_filter = np.array([[0, -1, 0],
                                   [-1, 5, -1],
                                   [0, -1, 0]], np.float32)

    # 3. 라플라시안 필터(3x3)
    laplacian_filter = np.array([[0, 1, 0],
                                  [1, -4, 1],
                                  [0, 1, 0]], np.float32)

    apply_filter(image, average_filter, 'Average Filter')      # 평균 값 필터 적용
    apply_filter(image, sharpening_filter, 'Sharpening Filter')  # 샤프닝 필터 적용
    apply_filter(image, laplacian_filter, 'Laplacian Filter')   # 라플라시안 필터 적용

    cv2.waitKey(0)
    cv2.destroyAllWindows()
```
